```python
#Task 01)Boxplot: Flipkart Product Price Distribution

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "product": [
        "Mobile", "Laptop", "Headphones", "TV", "Watch",
        "Camera", "Tablet", "Speaker", "Keyboard", "Mouse",
        "Mobile", "Laptop", "TV", "Camera", "Tablet"
    ],
    "price": [
        12000, 55000, 2500, 42000, 3500,
        30000, 18000, 4500, 1500, 800,
        15000, 60000, 45000, 28000, 20000
    ]
})


sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.boxplot(y=df["price"])

plt.title("Flipkart Product Price Distribution")
plt.ylabel("Price (₹)")

plt.tight_layout()
plt.show()
```

```python
# Task 02)Violinplot: Order Amounts, Fast Food vs Desserts
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "restaurant_category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts"
    ],
    "order_amount": [
        250, 300, 450, 350, 500,
        275, 400, 600, 325, 550,
        150, 200, 250, 300, 180,
        350, 220, 275, 400, 320
    ]
})


sns.set_theme(style="whitegrid")


plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="restaurant_category",
    y="order_amount"
)

plt.title("Order Amount Distribution: Fast Food vs Desserts")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")

plt.tight_layout()
plt.show()
```

```python
# Task 03) Boxplot: IPL Ticket Prices, Identify Outliers
import seaborn as sns
import matplotlib.pyplot as plt

# IPL match ticket prices
ticket_prices = [
    500, 750, 1000, 1200, 1500,
    1800, 2000, 2200, 2500, 3000,
    3500, 4000, 4500, 5000, 6000,
    7500, 8000, 9000, 10000, 12000,
    25000
]

# Set theme
sns.set_theme(style="whitegrid")

# Boxplot
plt.figure(figsize=(9, 5))

sns.boxplot(x=ticket_prices)

plt.title("IPL Match Ticket Price Distribution")
plt.xlabel("Ticket Price (₹)")

plt.tight_layout()
plt.show()
```

```python
# Task -04)Violinplot: Daily Step Counts (Google Fit-style)
import seaborn as sns
import matplotlib.pyplot as plt


steps = [
    3000, 4500, 5200, 6000, 6500,
    7000, 7200, 7500, 7800, 8000,
    8200, 8500, 8800, 9000, 9200,
    9500, 10000, 10500, 11000, 12000,
    5000, 6800, 7600, 8400, 9000,
    7300, 8100, 8700, 9800, 11500
]


sns.set_theme(style="whitegrid")


plt.figure(figsize=(8, 5))

sns.violinplot(y=steps)

plt.title("Distribution of Daily Step Counts")
plt.ylabel("Daily Steps")

plt.tight_layout()
plt.show()
```
