```python
#Task 01)orders per food category

import seaborn as sns
import matplotlib.pyplot as plt

# Data
categories = ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"]
orders = [450, 380, 520, 300, 250]

# Seaborn theme
sns.set_theme(style="whitegrid")

# Figure size
plt.figure(figsize=(10, 6))

# Barplot
sns.barplot(x=categories, y=orders)

# Title and labels
plt.title("Orders by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")

# Prevent overlap
plt.tight_layout()

# Save as PNG with 300 DPI
plt.savefig("food_category_orders.png", dpi=300, bbox_inches="tight")

# Display chart
plt.show()
```

```python
#Task 02)Line Plot Exported as SVG
import seaborn as sns
import matplotlib.pyplot as plt

# Data
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

steps = [4500, 5200, 6100, 5800, 7000, 6500, 7500]

# Seaborn theme
sns.set_theme(style="whitegrid")

# Create line plot
plt.figure(figsize=(10, 6))

sns.lineplot(x=days, y=steps, marker="o")

# Title and labels
plt.title("Daily Step Counts - One Week")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

# Prevent label overlap
plt.xticks(rotation=20)
plt.tight_layout()

# Save as SVG
plt.savefig("weekly_step_counts.svg", format="svg", bbox_inches="tight")

# Display
plt.show()
```

```python
# Task 03)Darkgrid Chart Exported as 1200 DPI PNG
import seaborn as sns
import matplotlib.pyplot as plt

# Set Seaborn theme
sns.set_theme(style="darkgrid", palette="deep")

# Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [120, 180, 150, 220, 300, 250, 200]

# Create barplot
plt.figure(figsize=(10, 6))

sns.barplot(x=days, y=sales)

# Title and labels
plt.title("Daily Sales")
plt.xlabel("Day")
plt.ylabel("Sales")

# Adjust layout
plt.tight_layout()

# Save as high-resolution PNG
plt.savefig("daily_sales_1200dpi.png", dpi=1200, bbox_inches="tight")

# Display
plt.show()
```

```python
# Task 04)apply_presentation_theme() Template Function + Pie Chart Export

import seaborn as sns
import matplotlib.pyplot as plt

# Template function
def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        context="talk",
        palette="deep",
        font_scale=1.1
    )

# Apply presentation theme
apply_presentation_theme()

# Top 5 most-used apps
apps = ["WhatsApp", "Instagram", "YouTube", "Facebook", "Spotify"]
usage = [35, 25, 20, 12, 8]

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    usage,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

# Keep pie chart circular
plt.axis("equal")

# Save as high-quality PNG
plt.savefig(
    "top_5_apps.png",
    dpi=300,
    bbox_inches="tight"
)

# Display chart
plt.show()
```

```python
#Task 05)AI-Generated Snippet: Exporting a Heatmap as 1200 DPI SVG
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# ChatGPT Prompt:
# "Generate a code snippet for exporting a Seaborn heatmap
# as a 1200 DPI SVG file."

# Code received from ChatGPT:
# plt.savefig("heatmap.svg", format="svg", dpi=1200)

# Sample data
data = np.array([
    [10, 20, 30, 40, 50],
    [15, 25, 35, 45, 55],
    [12, 22, 32, 42, 52],
    [18, 28, 38, 48, 58],
    [20, 30, 40, 50, 60]
])

# Create heatmap
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))

sns.heatmap(
    data,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Sales Heatmap")
plt.xlabel("Product")
plt.ylabel("Region")

plt.tight_layout()

# Export as 1200 DPI SVG
plt.savefig(
    "sales_heatmap_1200dpi.svg",
    format="svg",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```
