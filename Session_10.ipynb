```python
import seaborn as sns
import matplotlib.pyplot as plt

# Spotify data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
songs_streamed = [120, 180, 150, 220, 300, 250, 200]

# Apply poster context
sns.set(context="poster")

# Barplot
sns.barplot(x=days, y=songs_streamed)

plt.title("Spotify Songs Streamed Per Day")
plt.xlabel("Day")
plt.ylabel("Number of Songs")
plt.show()
```

```python
#observation: The poster context makes the plot elements, such as the title, labels, and text, larger and easier to read compared to the default context. It is useful for presentations and large displays.
```

```python
# Task 02)Countplot with 'colorblind' Palette: Food Orders by Cuisine
import seaborn as sns
import matplotlib.pyplot as plt

# Sample food order data
cuisine = [
    "Indian", "Chinese", "Italian", "Indian", "Chinese",
    "Indian", "Italian", "Chinese", "Indian", "Italian",
    "Indian", "Chinese", "Indian", "Italian", "Chinese"
]

# Create countplot
sns.countplot(x=cuisine, palette="colorblind")

plt.title("Food Orders by Cuisine Type")
plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Task 03)Lineplot: 2 Weeks of Steps, 'muted' Palette
import seaborn as sns
import matplotlib.pyplot as plt

# Simulated daily step counts for 2 weeks
days = [
    "Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7",
    "Day 8", "Day 9", "Day 10", "Day 11", "Day 12", "Day 13", "Day 14"
]

steps = [4500, 5200, 6100, 5800, 7000, 6500, 7200,
         6800, 7500, 8000, 7300, 8500, 8200, 9000]

# Apply muted palette and notebook context
sns.set_theme(context="notebook", palette="muted")

# Create lineplot
sns.lineplot(x=days, y=steps, marker="o")

plt.title("Daily Step Counts - 2 Weeks")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# task 04)Refactor: 'deep'→'colorblind' Palette, 'talk'→'poster' Context

import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
orders = [120, 150, 180, 140, 220, 250, 200]

# Original settings
sns.set_theme(context="talk", palette="deep")

# Refactored settings
sns.set_theme(context="poster", palette="colorblind")

# Create plot
sns.barplot(x=days, y=orders)

plt.title("Daily Orders")
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# Task 05)Flipkart-Dashboard-Style Theme: 'deep' Palette, 'talk' Context, Custom Font Scale
import seaborn as sns
import matplotlib.pyplot as plt

# Flipkart-style theme
sns.set(
    context="talk",
    palette="deep",
    font_scale=1.1,
    style="whitegrid"
)

# Data
categories = ["Mobiles", "Electronics", "Fashion", "Home", "Grocery"]
sales = [850, 650, 500, 400, 300]

# Increase figure size
plt.figure(figsize=(10, 6))

# Barplot
sns.barplot(x=categories, y=sales)

# Labels
plt.title("Flipkart Sales Dashboard")
plt.xlabel("Product Category")
plt.ylabel("Sales (₹ Lakhs)")

# Rotate category names
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```
