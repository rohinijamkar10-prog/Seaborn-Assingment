```python
# Task 01) Histplot: Swiggy Delivery Times (35 values)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "delivery_time": [
        25, 30, 35, 28, 32, 40, 45, 38, 27, 31,
        29, 36, 42, 33, 26, 34, 39, 41, 30, 28,
        35, 37, 43, 32, 29, 31, 40, 44, 36, 33,
        27, 38, 34, 30, 42
    ]
})

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="delivery_time",
    bins=8
)

plt.title("Distribution of Swiggy Delivery Times")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

```python
#Task 02)Kdeplot: One Month of Daily Step Counts
import seaborn as sns
import matplotlib.pyplot as plt

# Daily step counts for one month
steps = [
    4500, 5200, 6100, 7000, 6800,
    7500, 8200, 9000, 7800, 6500,
    5900, 7200, 8100, 8500, 9200,
    10000, 8800, 7600, 6900, 7300,
    6400, 5700, 7100, 8000, 8700,
    9500, 10200, 8900, 7700, 6800
]

# Set Seaborn theme
sns.set_theme(style="whitegrid")

# Create KDE plot
plt.figure(figsize=(8, 5))

sns.kdeplot(
    x=steps,
    fill=True
)

plt.title("Daily Step Count Distribution")
plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")

plt.tight_layout()
plt.show()
```

```python
#3)Histplot + Kdeplot Overlay: Zomato Monthly Spending
import seaborn as sns
import matplotlib.pyplot as plt


spending = [
    250, 400, 550, 700, 850,
    900, 1000, 1200, 1350, 1500,
    1600, 1750, 1800, 1950, 2100,
    2200, 2400, 2500, 2700, 2900,
    3000, 3200, 3500, 3700, 4000,
    4200, 4500, 4800, 5000, 5500
]


sns.set_theme(style="whitegrid")


fig, ax = plt.subplots(figsize=(9, 5))


sns.histplot(
    spending,
    bins=8,
    stat="density",
    alpha=0.5,
    ax=ax
)


sns.kdeplot(
    spending,
    ax=ax
)


ax.set_title("Distribution of Monthly Zomato Spending")
ax.set_xlabel("Monthly Spending (₹)")
ax.set_ylabel("Density")

plt.tight_layout()
plt.show()
```
