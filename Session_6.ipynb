```python
# Task 01)Jointplot (scatter): Flipkart Rating vs Price


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample Flipkart product data
df = pd.DataFrame({
    "product": [
        "Mobile", "Laptop", "Headphones", "TV", "Smartwatch",
        "Camera", "Tablet", "Speaker", "Keyboard", "Mouse",
        "Mobile", "Laptop", "Headphones", "TV", "Smartwatch"
    ],
    "rating": [
        4.5, 4.2, 4.0, 4.6, 4.1,
        4.4, 3.9, 4.3, 4.0, 3.8,
        4.7, 4.5, 4.2, 4.3, 4.0
    ],
    "price": [
        15000, 55000, 2500, 45000, 3500,
        30000, 18000, 4000, 1500, 800,
        17000, 60000, 3000, 48000, 4000
    ]
})

# Create jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="price",
    kind="scatter"
)

plt.show()
```

```python
#4)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Zomato restaurant review data
df = pd.DataFrame({
    "rating": [
        2.5, 2.8, 3.0, 3.1, 3.2, 3.3, 3.4, 3.5, 3.5, 3.6,
        3.7, 3.7, 3.8, 3.8, 3.9, 3.9, 4.0, 4.0, 4.1, 4.1,
        4.2, 4.2, 4.3, 4.3, 4.4, 4.4, 4.5, 4.5, 4.6, 4.7
    ],
    "number_of_reviews": [
        20, 35, 50, 70, 90, 120, 150, 180, 220, 250,
        300, 350, 400, 450, 500, 600, 700, 800, 950, 1100,
        1300, 1500, 1800, 2000, 2300, 2600, 3000, 3500, 4000, 4500
    ]
})

# 1. Scatter Jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="number_of_reviews",
    kind="scatter"
)
plt.show()


# 2. Hexbin Jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="number_of_reviews",
    kind="hex"
)
plt.show()


# 3. KDE Jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="number_of_reviews",
    kind="kde"
)
plt.show()
```

```python
#The scatter plot shows individual restaurants and helps identify the relationship between rating and number of reviews, but points may overlap when the dataset is large. The hexbin plot shows the concentration of restaurants in different rating and review ranges, making dense areas easier to identify. The KDE plot provides a smooth view of the overall data density but does not show individual observations clearly. Among the three, the hexbin plot is the most useful because it clearly shows where the data is highly concentrated, especially when many points overlap.
```
