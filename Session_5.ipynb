```python
#Task 01) Basic Pairplot: All Numerical Columns

import seaborn as sns
import matplotlib.pyplot as plt

# Load tips dataset
tips = sns.load_dataset("tips")

# Create pairplot for all numerical columns
sns.pairplot(tips)

plt.show()
```

```python
# Task 02)Pairplot with hue='sex'
import seaborn as sns
import matplotlib.pyplot as plt

# Load tips dataset
tips = sns.load_dataset("tips")

# Pairplot with hue = sex
sns.pairplot(tips, hue="sex")

plt.show()
```

```python
# Task 03)Pairplot with hue='smoker' and diag_kind='kde'
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.pairplot(
    tips,
    hue="smoker",
    diag_kind="kde"
)

plt.show()
```

```python
# Task 04)Pairplot: Selected Columns Only, hue='time'
import seaborn as sns
import matplotlib.pyplot as plt


tips = sns.load_dataset("tips")

sns.pairplot(
    tips,
    vars=["total_bill", "tip"],
    hue="time"
)

plt.show()
```
