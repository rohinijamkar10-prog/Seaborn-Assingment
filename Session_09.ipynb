```python
# Task 01)Regplot: Instagram Likes vs Comments (10 Posts)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Instagram data
df = pd.DataFrame({
    "likes": [100, 200, 350, 500, 700, 900, 1200, 1500, 1800, 2200],
    "comments": [10, 18, 25, 35, 45, 55, 70, 85, 100, 120]
})

# Regression plot
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="likes",
    y="comments"
)

plt.title("Instagram Likes vs Comments")
plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")

plt.tight_layout()
plt.show()
```

```python
# Task 02)Lmplot: Order Price vs Delivery Time (15 Zomato-style Rows)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Zomato-style dataset
df = pd.DataFrame({
    "price": [
        120, 180, 250, 300, 350,
        420, 500, 550, 650, 700,
        800, 900, 1000, 1200, 1500
    ],
    "delivery_time": [
        20, 22, 25, 28, 30,
        32, 35, 37, 40, 42,
        45, 48, 50, 55, 60
    ]
})

# Create lmplot
sns.set_theme(style="whitegrid")

sns.lmplot(
    data=df,
    x="price",
    y="delivery_time",
    height=5,
    aspect=1.5
)

plt.title("Food Order Price vs Delivery Time")
plt.xlabel("Order Price (₹)")
plt.ylabel("Delivery Time (Minutes)")

plt.show()
```

```python
#Task 03)Regplot with Lowess Smoothing: Playlist Songs

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Spotify playlist data
df = pd.DataFrame({
    "songs": [10, 15, 20, 25, 30, 35, 40, 45, 50,
              55, 60, 65, 70, 75, 80],
    "duration": [35, 50, 65, 75, 95, 110, 125, 145, 155,
                 175, 190, 205, 225, 240, 260]
})

# Lowess smoothing
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="songs",
    y="duration",
    lowess=True
)

plt.title("Number of Songs vs Playlist Duration")
plt.xlabel("Number of Songs")
plt.ylabel("Total Playlist Duration (Minutes)")

plt.tight_layout()
plt.show()
```

```python
# Task 04)Faceted Lmplot: YouTube Hours vs Exam Scores by Preparation Method

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours_youtube": [2, 4, 5, 7, 8, 3, 5, 6, 9, 10],
    "exam_score": [55, 62, 65, 72, 78, 58, 67, 70, 82, 88],
    "preparation_method": [
        "self-study", "self-study", "self-study", "self-study", "self-study",
        "coaching", "coaching", "coaching", "coaching", "coaching"
    ]
})

sns.lmplot(
    data=df,
    x="hours_youtube",
    y="exam_score",
    col="preparation_method",
    height=5,
    aspect=1
)

plt.show()
```

```python
# Task 05)Given Lmplot: Flipkart Price vs Reviews (Aesthetics Only, Same Regression Logic)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Flipkart product data
df = pd.DataFrame({
    "price": [499, 799, 999, 1299, 1599, 1999, 2499, 2999, 3499, 3999,
              4499, 4999, 5499, 5999, 6499],

    "reviews": [1200, 1500, 1800, 2100, 2500, 2800, 3200, 3500, 3900, 4200,
                4500, 4800, 5200, 5500, 6000]
})

# Create regression plot
g = sns.lmplot(
    data=df,
    x="price",
    y="reviews",
    scatter_kws={"color": "red"}
)

# Customize aesthetics only
g.set_axis_labels("Product Price (₹)", "Number of Reviews")
g.fig.suptitle("Flipkart Product Price vs Number of Reviews")

# Adjust title position
g.fig.subplots_adjust(top=0.90)

plt.show()
```
