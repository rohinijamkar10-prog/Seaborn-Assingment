```python
#Task 3 — Axes-Level vs Figure-Level: lineplot() vs relplot()
#Axes-level function — sns.lineplot()

import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend')

plt.show()
```

```python
#3)2)2. Figure-level function — sns.relplot()
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.relplot(
    x=days,
    y=temps,
    kind='line',
    marker='o'
)

plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend')

plt.show()
```

```python
#Difference to  Graph

#Axes-level (sns.lineplot()) works directly on a specific Matplotlib Axes, so it gives more control over an individual plot.

#Figure-level (sns.relplot()) creates and manages the entire figure, making it easier to create multiple plots using options such as row, col, or hue.
```

```python
#Task 04) Applying the Three Built-in Themes

import seaborn as sns
import matplotlib.pyplot as plt

# Temperature data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [28, 30, 29, 32, 34, 33, 31]

# Themes to apply
themes = ["darkgrid", "whitegrid", "ticks"]

for theme in themes:
    # Apply Seaborn theme
    sns.set_theme(style=theme)

    # Create plot
    plt.figure(figsize=(8, 5))
    sns.lineplot(x=days, y=temperature, marker="o")

    plt.title(f"Temperature Trend - {theme}")
    plt.xlabel("Day")
    plt.ylabel("Temperature (°C)")

    # Save separate image
    plt.savefig(f"temperature_{theme}.png", dpi=150, bbox_inches="tight")

    # Display plot
    plt.show()

    # Close figure
    plt.close()
```

```python
#darkgrid theme gives a darker background with prominent grid lines, making the chart easy to read. Whitegrid keeps the background white while still showing grid lines, giving a clean and professional look. Ticks removes the grid and provides a simpler, minimal appearance with only axis tick marks.
```
