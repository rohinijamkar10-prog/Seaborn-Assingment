```python
# Task 01)DataFrame + Correlation Matrix: Food Delivery Order Features
import pandas as pd
import numpy as np

# Random data
np.random.seed(42)

df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 60, 20),
    "order_value": np.random.randint(100, 1000, 20),
    "tip_amount": np.random.randint(10, 150, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 1)
})

# Display DataFrame
print("Food Delivery Data:")
print(df)

# Calculate correlation matrix
correlation_matrix = df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
# Task 02)Seaborn Heatmap with Annotations
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Create food delivery data
np.random.seed(42)

df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 60, 20),
    "order_value": np.random.randint(100, 1000, 20),
    "tip_amount": np.random.randint(10, 150, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 1)
})

# Calculate correlation matrix
corr_matrix = df.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap - Food Delivery Data")
plt.tight_layout()
plt.show()
```

```python
# Task 03)Diverging Color Palette (coolwarm)
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Create food delivery data
np.random.seed(42)

df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 60, 20),
    "order_value": np.random.randint(100, 1000, 20),
    "tip_amount": np.random.randint(10, 150, 20),
    "rating": np.round(np.random.uniform(3.0, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 1)
})

# Correlation matrix
corr_matrix = df.corr()

# Heatmap with diverging palette
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap - Food Delivery Data")
plt.tight_layout()
plt.show()

# Find strongest negative correlation
corr_pairs = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
).stack()

strongest_negative = corr_pairs[corr_pairs < 0].idxmin()
strongest_value = corr_pairs.min()

print("Strongest Negative Correlation:")
print(strongest_negative)
print("Correlation:", round(strongest_value, 2))
```

```python
# Task 04)Flipkart Mobile Phones: Correlation Heatmap

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Small Flipkart Mobile Phones dataset
df = pd.DataFrame({
    "price": [12000, 15000, 18000, 22000, 25000, 30000, 35000, 40000, 45000, 50000,
              14000, 20000, 28000, 32000, 38000],

    "rating": [4.1, 4.2, 4.0, 4.3, 4.4, 4.5, 4.2, 4.6, 4.3, 4.5,
               4.1, 4.0, 4.4, 4.5, 4.2],

    "RAM": [4, 4, 6, 6, 8, 8, 8, 12, 12, 12,
            4, 6, 8, 8, 12],

    "battery_capacity": [4000, 4500, 5000, 5000, 4500, 5000, 5000, 6000, 5000, 6000,
                         4000, 4500, 5000, 5000, 6000]
})

# Display dataset
print("Flipkart Mobile Data:")
print(df)

# Calculate correlation matrix
corr_matrix = df.corr()

print("\nCorrelation Matrix:")
print(corr_matrix)

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Flipkart Mobile Phones Correlation Heatmap")
plt.tight_layout()
plt.show()
```
